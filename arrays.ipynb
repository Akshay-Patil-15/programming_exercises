```python
# 1929. Concatenation of Array.
# https://leetcode.com/problems/concatenation-of-array/description/
class Solution:
    def getConcatenation(self, nums: List[int]) -> List[int]:
        result = [0] * 2 *len(nums)
        for i, ele in enumerate(nums):
            result[i], result[i + len(nums)] = ele, ele
        return result
        # TC, SC = O(n), O(n).
```

```python
# 217. Contains duplicate.
# https://leetcode.com/problems/contains-duplicate/description/
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        ele_set = set()
        for ele in nums:
            if ele in ele_set: return True # Duplicate encountered.
            ele_set.add(ele)
        return False
        # TC, SC = O(n), O(n).
```

```python
# 242. Valid Anagram.
# https://leetcode.com/problems/valid-anagram/
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        # Method 1: Inbuilt Counter function.
        # return Counter(s) == Counter(t)
        # TC, SC = O(n), O(n).

        # Method 2: Creating dictionaries manually.
        # s_dict, t_dict = {}, {}
        # for ele in s:
        #     s_dict[ele] = 1 + s_dict.get(ele, 0)
        # for ele in t:
        #     t_dict[ele] = 1 + t_dict.get(ele, 0)
        # return s_dict == t_dict
        # TC, SC = O(n), O(n).

        # Method 3: Constant space using frequency list.
        if len(s) != len(t): return False # Egde case.
        
        frequency_list = [0]*26
        for ele in s:
            frequency_list[ord(ele) - ord("a")] += 1
        for ele in t:
            if frequency_list[ord(ele) - ord("a")] == 0: return False
            frequency_list[ord(ele) - ord("a")] -= 1
        return True
        # TC = O(n)
        # SC = O(26) = O(1).
```

```python
# 1. Two Sum
# https://leetcode.com/problems/two-sum/description/
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        nums_dict = {} # element : index
        for i, ele in enumerate(nums):
            if target - ele in nums_dict: return [i, nums_dict[target - ele]]
            nums_dict[ele] = i
        # TC, SC = O(n), O(n).
```
