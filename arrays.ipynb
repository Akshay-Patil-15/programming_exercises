```python
# 1929. Concatenation of Array
class Solution:
    def getConcatenation(self, nums: List[int]) -> List[int]:
        result = [0] * 2 *len(nums)
        for i, ele in enumerate(nums):
            result[i], result[i + len(nums)] = ele, ele
        return result
        # TC, SC = O(n), O(n).
```
