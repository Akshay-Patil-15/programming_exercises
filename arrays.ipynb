```python
from typing import List
```

```python
# 1929. Concatenation of Array.

class Solution:
    def getConcatenation(self, nums: List[int]) -> List[int]:
        result = [0] * 2 *len(nums)
        for i, ele in enumerate(nums):
            result[i], result[i + len(nums)] = ele, ele
        return result
        # TC, SC = O(n), O(n).
```

```python
# 217. Contains duplicate.
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        ele_set = set()
        for ele in nums:
            if ele in ele_set: return True # Duplicate encountered.
            ele_set.add(ele)
        return False
        # TC, SC = O(n), O(n).
```
